## 4. Test Script Execution (simulates the Task Scheduler)
This cell runs your script the same way the scheduler does — in a subprocess with captured output.

```python
import subprocess, sys, tempfile, os

def test_as_scheduled_task(script_code, timeout=60):
    """Simulate how the task scheduler executes your script."""
    fd, path = tempfile.mkstemp(suffix='.py', prefix='task_test_')
    try:
        with os.fdopen(fd, 'w') as f:
            f.write(script_code)
        result = subprocess.run(
            [sys.executable, path],
            capture_output=True, text=True, timeout=timeout
        )
        print('=== STATUS:', 'SUCCESS' if result.returncode == 0 else 'FAILURE', '===')
        if result.stdout:
            print('--- Output ---')
            print(result.stdout)
        if result.stderr:
            print('--- Errors ---')
            print(result.stderr)
    except subprocess.TimeoutExpired:
        print(f'TIMEOUT: Script took longer than {timeout}s')
    finally:
        os.unlink(path)

# Paste your script as a string to test it:
test_as_scheduled_task(r"""
from datetime import datetime
print(f'Hello from scheduled task at {datetime.now()}')
""")
```
